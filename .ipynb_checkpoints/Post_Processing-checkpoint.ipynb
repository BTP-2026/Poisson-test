```python
import os
os.environ['CUDA_VISIBLE_DEVICES'] = '0'
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as tick
import matplotlib as mpl
import h5py
import os
from utils import get_train_data, read_h5_file

get_dir = os.getcwd()
os.chdir(get_dir)
```

```python
def get_predictions_and_true_data_poisson(mdir,ddir, sen_dir, indices):
    sen_ind = pd.read_csv(sen_dir)['sensor_indices'].values.reshape(-1)
    model = tf.keras.models.load_model(mdir, safe_mode=False)
    seq_len = len(sen_ind)
    # Read Poisson data using the new utility function
    f_all, xdisc, u_all, bc_all = read_h5_file(ddir)
    X = np.repeat(np.expand_dims(xdisc, axis=0), 100, axis=0)
    x_sen = np.repeat(X[0, sen_ind].reshape((1, 1, -1)), 100, axis=0)
    print(x_sen.shape)
    ins = X.shape[1]
    print(ins)
    u_pred = []
    u_true = []
    err = []
    f = []
    xs = []
   
    for i in indices:
        f.append(f_all[i].reshape(-1, 1))
        xs.append(xdisc)
        u_sen = u_all[i, sen_ind].reshape((-1, seq_len, 1))
        xbc = np.repeat(x_sen[i], len(xdisc), axis=1).reshape((-1, seq_len))
        ubc = np.repeat(u_sen, len(xdisc), axis=1).reshape((-1, seq_len))

        u_pred.append(model.predict([xdisc, xbc, ubc], verbose=0).reshape(-1))
        u_true.append(u_all[i].reshape(-1))
    
    return xs, f, u_pred, u_true
```

```python
def plot_poisson_solutions(x, u_pred, u_true, f, indices, err, plot_save=False, plot_dir='poisson_results', fig_size=(15, 10)):
    n_samples = len(indices)
    print(n_samples)
    fig, ax = plt.subplots(4, n_samples, figsize=fig_size, sharex=True)
    
    
    for i in range(n_samples):
        # Solution p(x)
        ax[0][i].plot(x[i], u_pred[i], 'b-', label='PINTO Prediction', linewidth=2)
        ax[0][i].plot(x[i], u_true[i], 'r--', label='True Solution', linewidth=2)
        ax[0][i].set_title(f'Solution p(x) - Sample {indices[i]}')
        ax[0][i].set_ylabel('u(x)')
        ax[0][i].legend()
        ax[0][i].grid(True, alpha=0.3)
        
        # Forcing term f(x)
        ax[2][i].plot(x[i], f[i], 'm-', label='Forcing f(x)', linewidth=2)
        ax[2][i].set_title(f'Forcing Term f(x) - Sample {indices[i]}')
        ax[2][i].set_ylabel('f(x)')
        ax[2][i].legend()
        ax[2][i].grid(True, alpha=0.3)
        
        # Relative error
        # rel_error = err[i]
        # ax[3][i].plot(x[i], rel_error, 'k-', label='Relative Error', linewidth=2)
        # ax[3][i].set_title(f'Relative Error - Sample {indices[i]}')
        # ax[3][i].set_ylabel('Relative Error')
        # ax[3][i].set_xlabel('x')
        # ax[3][i].legend()
        # ax[3][i].grid(True, alpha=0.3)
    
    plt.tight_layout()
    
    if plot_save:
        plt.savefig(f'{plot_dir}.png', dpi=300, bbox_inches='tight', format='png')
    
    plt.show()
```

```python
# Set up paths and parameters
mdir = './Poisson_PINTO/Poisson_model.keras'  # Path to trained model
ddir = './data/poisson_5000.h5'  # Path to Poisson data
sen_dir = './Poisson_PINTO/sensor.csv'  # Path to sensor indices CSV

# Define sample indices for analysis
train_indices = [10, 25, 40, 55]  # Seen boundary conditions
test_indices = [85, 90, 95]       # Unseen boundary conditions
all_indices = train_indices + test_indices

# Generate predictions
xs, f, u_pred, u_true = get_predictions_and_true_data_poisson(
    mdir, ddir, sen_dir=sen_dir, indices=all_indices 
)

plot_poisson_solutions(xs[:len(train_indices)], u_pred[:len(train_indices)], u_true[:len(train_indices)], f[:len(train_indices)], train_indices, [], plot_save=True, plot_dir='Poisson_Train_Results', fig_size=(16, 12))
```
